# Step 2: Color Space Conversions - RGB, Grayscale, and Binary

## RGB (Red, Green, Blue) Color Model
- **Definition**: RGB is an additive color model where any color is created by combining red, green, and blue light in different intensities.
- **Channels**: 3 channels (Red, Green, Blue), each with values from 0-255.
- **Use cases**: Natural color image representation, displays, photography, and most digital imaging applications.
- **Characteristics**: Intuitive for human perception and widely supported by imaging hardware.

## Grayscale Image
- **Definition**: Grayscale (or greyscale) is a color model where each pixel is represented by a single intensity value.
- **Channels**: 1 channel with values from 0-255 (0 = black, 255 = white).
- **Conversion**: Typically computed as a weighted average of RGB channels: Gray = 0.299*R + 0.587*G + 0.114*B
- **Use cases**: Image processing algorithms, edge detection, simple storage, medical imaging, and when color information is not needed.
- **Advantages**: Reduces memory usage and computational complexity.

## Binary Image
- **Definition**: A binary image contains only two intensity values: black (0) and white (255).
- **Channels**: 1 channel with only values 0 or 255.
- **Conversion**: Usually created by applying a threshold to a grayscale image. Pixels above threshold become white (255), below become black (0).
- **Use cases**: Segmentation, document scanning, silhouette extraction, and applications requiring clear distinction between foreground and background.
- **Advantages**: Simplest form of image representation with minimum storage and fast processing.

# Step 3: HSV Color Model - Explanation and Conversion

## HSV (Hue, Saturation, Value) Color Model

**Definition**: HSV is a cylindrical color model that represents colors in a more intuitive way compared to RGB. It separates color information (hue) from intensity and saturation.

**Three Components:**

1. **Hue (H)**: Represents the color type
   - Range: 0-180Â° (in OpenCV, 0-180 to fit in uint8)
   - 0Â° = Red, 60Â° = Yellow, 120Â° = Green, 180Â° = Cyan, etc.
   - Defines the color itself independent of brightness or saturation

2. **Saturation (S)**: Represents the color intensity/purity
   - Range: 0-255 (0-100%)
   - 0 = Pure gray (no color), 255 = Pure color
   - Lower values appear more grayish, higher values are more vivid

3. **Value (V)**: Represents the brightness
   - Range: 0-255 (0-100%)
   - 0 = Black, 255 = Maximum brightness
   - V is sometimes called "Lightness" or "Brightness"

**Why HSV is Useful:**
- More intuitive for color selection and manipulation
- Better for color-based image segmentation (isolating specific colors)
- Separates brightness from color, useful for lighting-invariant applications
- Easier to define color ranges (e.g., "all red hues")
- Popular in computer vision for object detection and color tracking

# Step 5: Image Contrast Analysis and Manipulation

## Contrast Definition

**Contrast** is a measure of the difference in brightness between the lighter and darker areas of an image. It represents the range of tonal variations in an image.

**Formal Definition**: Contrast can be calculated as:
- **Standard Deviation Method**: Contrast = Standard Deviation of pixel values
- **Min-Max Method**: Contrast = (Max Pixel Value - Min Pixel Value) / (Max Pixel Value + Min Pixel Value)
- **Michelson Contrast**: Contrast = (L_max - L_min) / (L_max + L_min), where L_max and L_min are maximum and minimum luminance

**Interpretation:**
- **High Contrast**: Large difference between bright and dark areas;
- **Low Contrast**: Small difference;

**Applications:**
- Image quality assessment
- Preprocessing for object detection
- Enhancement for better visibility
- Medical imaging analysis

```python
def calculate_contrast(image):
    """Calculate contrast as standard deviation of pixel values"""
    return np.std(image)

original_contrast = calculate_contrast(image_gray)

print("=== Contrast Calculation ===")
print(f"Original image contrast (std): {original_contrast:.2f}")
print(f"Original image min value: {image_gray.min()}")
print(f"Original image max value: {image_gray.max()}")
print(f"Original image mean value: {image_gray.mean():.2f}")

clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
image_gray_enhanced = clahe.apply(image_gray)
enhanced_contrast = calculate_contrast(image_gray_enhanced)

def adjust_contrast(image, factor):
    """Adjust contrast by a given factor"""
    adjusted = np.clip((image.astype(np.float32) - 128) * factor + 128, 0, 255).astype(np.uint8)
    return adjusted

image_gray_increased = adjust_contrast(image_gray, 1.5)
increased_contrast = calculate_contrast(image_gray_increased)

image_gray_decreased = adjust_contrast(image_gray, 0.5)
decreased_contrast = calculate_contrast(image_gray_decreased)

print(f"\nAfter CLAHE enhancement, contrast (std): {enhanced_contrast:.2f}")
print(f"After manual increase (factor=1.5), contrast (std): {increased_contrast:.2f}")
print(f"After manual decrease (factor=0.5), contrast (std): {decreased_contrast:.2f}")

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

axes[0, 0].imshow(image_gray, cmap='gray')
axes[0, 0].set_title(f"Original Image\nContrast (Std): {original_contrast:.2f}")
axes[0, 0].axis('off')

axes[0, 1].imshow(image_gray_enhanced, cmap='gray')
axes[0, 1].set_title(f"Enhanced Contrast (CLAHE)\nContrast (Std): {enhanced_contrast:.2f}")
axes[0, 1].axis('off')

axes[1, 0].imshow(image_gray_increased, cmap='gray')
axes[1, 0].set_title(f"Increased Contrast (Factor=1.5)\nContrast (Std): {increased_contrast:.2f}")
axes[1, 0].axis('off')

axes[1, 1].imshow(image_gray_decreased, cmap='gray')
axes[1, 1].set_title(f"Decreased Contrast (Factor=0.5)\nContrast (Std): {decreased_contrast:.2f}")
axes[1, 1].axis('off')

plt.suptitle("Contrast Manipulation Comparison", fontsize=16, y=1.00)
plt.tight_layout()
plt.show()
```

# Step 6: Image Histogram Definition and Visualization

## Histogram Definition

**Histogram** is a graphical representation of the distribution of pixel intensity values in an image.

**Key Components:**
- **X-axis**: Pixel intensity values (0-255 for 8-bit images)
- **Y-axis**: Number of pixels (frequency or count) at each intensity level
- **Width**: Each bar represents a range of intensity values (usually 1 unit wide)
- **Height**: The frequency of pixels with that intensity value

**Characteristics:**
- **Peak locations**: Indicate the dominant brightness levels in the image
- **Spread**: Wide histogram indicates high contrast; narrow histogram indicates low contrast
- **Distribution shape**: Tells us about the image characteristics:
  - Left-skewed: Image is mostly dark
  - Right-skewed: Image is mostly bright
  - Centered: Image has balanced tones
  - Multiple peaks: Image has distinct brightness regions

**Applications:**
- Image quality assessment
- Exposure analysis
- Preprocessing for image enhancement
- Thresholding and segmentation
- Histogram equalization for contrast enhancement

# Step 7: Histogram Equalization and Histogram Stretching

## Histogram Equalization

**Definition**: Histogram equalization is a technique that redistributes pixel intensity values to produce a more uniform distribution across the entire range (0-255). It enhances contrast by spreading out the most frequent intensity values.

**How it works:**
1. Calculate the histogram of the image
2. Compute the cumulative distribution function (CDF)
3. Normalize the CDF
4. Map each pixel intensity to a new value based on the CDF
5. Result: A new image with more uniform intensity distribution

**Advantages:**
- Automatically enhances contrast
- Improves visibility of details in dark or bright regions
- No parameters to adjust (fully automatic)
- Works well for images with poor contrast

**Disadvantages:**
- Can produce unnatural results
- May over-enhance noise
- Not suitable for images that already have good contrast


## Histogram Stretching

**Definition**: Histogram stretching (or normalization) is a technique that expands the range of pixel intensities to use the full available range (0-255). It redistributes pixel values linearly to stretch from the minimum to maximum values found in the image.

**How it works:**
1. Find the minimum and maximum pixel values in the image
2. Apply linear transformation: new_pixel = (pixel - min) / (max - min) * 255
3. Result: Image using the full intensity range

**Advantages:**
- Simple and intuitive
- Preserves image characteristics better than equalization
- Provides more natural-looking results
- Good for images with limited range due to poor lighting

**Disadvantages:**
- Less aggressive contrast enhancement than equalization
- Sensitive to outliers (extreme dark or bright pixels)
- May not work well if image already spans full range

**Difference between Histogram Equalization and Stretching:**
- **Equalization**: Creates a uniform distribution (flat histogram)
- **Stretching**: Linearly expands the current distribution to use full range

```python
# Histogram Equalization
image_equalized = cv2.equalizeHist(image_gray)

# Histogram Stretching (Manual implementation)
def histogram_stretch(image):
    """Apply histogram stretching (normalization)"""
    min_val = image.min()
    max_val = image.max()
    stretched = ((image.astype(np.float32) - min_val) / (max_val - min_val) * 255).astype(np.uint8)
    return stretched

image_stretched = histogram_stretch(image_gray)

# Calculate histograms for comparison
hist_original = cv2.calcHist([image_gray], [0], None, [256], [0, 256])
hist_equalized = cv2.calcHist([image_equalized], [0], None, [256], [0, 256])
hist_stretched = cv2.calcHist([image_stretched], [0], None, [256], [0, 256])

# Display images and their histograms
fig = plt.figure(figsize=(16, 10))
gs = fig.add_gridspec(3, 3, hspace=0.3, wspace=0.3)

# Original
ax1 = fig.add_subplot(gs[0, 0])
ax1.imshow(image_gray, cmap='gray')
ax1.set_title("Original Image")
ax1.axis('off')

ax2 = fig.add_subplot(gs[0, 1])
ax2.imshow(image_equalized, cmap='gray')
ax2.set_title("Histogram Equalization")
ax2.axis('off')

ax3 = fig.add_subplot(gs[0, 2])
ax3.imshow(image_stretched, cmap='gray')
ax3.set_title("Histogram Stretching")
ax3.axis('off')

# Histograms - Line plots
ax4 = fig.add_subplot(gs[1, 0])
ax4.plot(hist_original, color='black', linewidth=1)
ax4.fill_between(range(256), hist_original.flatten(), alpha=0.3, color='gray')
ax4.set_title("Original Histogram")
ax4.set_xlabel("Intensity")
ax4.set_ylabel("Frequency")
ax4.grid(True, alpha=0.3)
ax4.set_xlim([0, 255])

ax5 = fig.add_subplot(gs[1, 1])
ax5.plot(hist_equalized, color='blue', linewidth=1)
ax5.fill_between(range(256), hist_equalized.flatten(), alpha=0.3, color='lightblue')
ax5.set_title("Equalized Histogram")
ax5.set_xlabel("Intensity")
ax5.set_ylabel("Frequency")
ax5.grid(True, alpha=0.3)
ax5.set_xlim([0, 255])

ax6 = fig.add_subplot(gs[1, 2])
ax6.plot(hist_stretched, color='green', linewidth=1)
ax6.fill_between(range(256), hist_stretched.flatten(), alpha=0.3, color='lightgreen')
ax6.set_title("Stretched Histogram")
ax6.set_xlabel("Intensity")
ax6.set_ylabel("Frequency")
ax6.grid(True, alpha=0.3)
ax6.set_xlim([0, 255])

# Cumulative histograms for comparison
ax7 = fig.add_subplot(gs[2, :])
cumulative_original = np.cumsum(hist_original) / np.cumsum(hist_original).max()
cumulative_equalized = np.cumsum(hist_equalized) / np.cumsum(hist_equalized).max()
cumulative_stretched = np.cumsum(hist_stretched) / np.cumsum(hist_stretched).max()

ax7.plot(cumulative_original, color='black', linewidth=2, label='Original', linestyle='-', alpha=0.7)
ax7.plot(cumulative_equalized, color='blue', linewidth=2, label='Equalized', linestyle='--', alpha=0.7)
ax7.plot(cumulative_stretched, color='green', linewidth=2, label='Stretched', linestyle='-.', alpha=0.7)
ax7.set_title("Cumulative Distribution Comparison")
ax7.set_xlabel("Intensity")
ax7.set_ylabel("Cumulative Probability")
ax7.grid(True, alpha=0.3)
ax7.legend(loc='upper left')
ax7.set_xlim([0, 255])
ax7.set_ylim([0, 1.05])

plt.suptitle("Histogram Equalization vs Histogram Stretching", fontsize=16, y=0.995)
plt.show()

print("=== Comparison Statistics ===\n")
print("Original Image:")
print(f"  Min intensity: {image_gray.min()}")
print(f"  Max intensity: {image_gray.max()}")
print(f"  Mean intensity: {image_gray.mean():.2f}")
print(f"  Std deviation: {np.std(image_gray):.2f}")

print("\nHistogram Equalized Image:")
print(f"  Min intensity: {image_equalized.min()}")
print(f"  Max intensity: {image_equalized.max()}")
print(f"  Mean intensity: {image_equalized.mean():.2f}")
print(f"  Std deviation: {np.std(image_equalized):.2f}")

print("\nHistogram Stretched Image:")
print(f"  Min intensity: {image_stretched.min()}")
print(f"  Max intensity: {image_stretched.max()}")
print(f"  Mean intensity: {image_stretched.mean():.2f}")
print(f"  Std deviation: {np.std(image_stretched):.2f}")
```

# Step 2: Salt & Pepper + Gaussian Noise 

- Salt-and-pepper noise randomly replaces pixels with black or white values.

- Gaussian noise adds random intensity variations following a normal distribution, commonly caused by sensor noise.

```python
def add_salt_pepper(image, prob=0.02):
    noisy = np.copy(image)

    salt = np.random.random(image.shape[:2]) < prob/2
    pepper = np.random.random(image.shape[:2]) < prob/2

    noisy[salt] = 255
    noisy[pepper] = 0

    return noisy


def add_gaussian_noise(image, mean=0, sigma=25):
    gauss = np.random.normal(mean, sigma, image.shape)
    noisy = image + gauss
    return np.clip(noisy, 0, 255).astype(np.uint8)


sp_noisy = add_salt_pepper(img)
gauss_noisy = add_gaussian_noise(img)

plt.figure(figsize=(18,6))

plt.subplot(1,3,1)
plt.imshow(img)
plt.title("Original")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(sp_noisy)
plt.title("Salt & Pepper Noise")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(gauss_noisy)
plt.title("Gaussian Noise")
plt.axis("off")

plt.show()
```

# Step 3 : Kernel explanation 

## Kernel in Image Processing

A kernel is a small matrix used to process images through convolution.

Applications:
- Noise reduction
- Edge detection
- Sharpening
- Blurring
- Feature extraction

The kernel slides over the image. At each position, corresponding values are multiplied and summed to generate a new pixel value.

# Step 4 : Custom Filtering Function

### Function Explanation : 
- Input image as a NumPy array. (h,w,c)
- Get kernel dimensions, using kh, kw
- Compute padding size, Padding is needed so filtering can be applied at border pixels. For a 3×3 kernel: `pad_h = 1` , `pad_w = 1` This means one pixel is added around the image.
- Pad the image. Without padding, edge pixels would not have enough neighbors for filtering. `mode='reflect'` means Border pixels are mirrored.
- Then we will loop through the immage, filling the values in the output we first made with the same image dimensions.

```python
def apply_kernel(image, kernel):
    h, w, c = image.shape
    kh, kw = kernel.shape

    pad_h = kh // 2
    pad_w = kw // 2

    padded = np.pad(
        image,
        ((pad_h,pad_h),(pad_w,pad_w),(0,0)),
        mode='reflect'
    )

    output = np.zeros_like(image)

    for y in range(h):
        for x in range(w):
            for ch in range(c):
                region = padded[y:y+kh, x:x+kw, ch]
                output[y,x,ch] = np.clip(
                    np.sum(region * kernel),
                    0,
                    255
                )

    return output.astype(np.uint8)
```

# Step 5: Low-Pass filters on the noisy image 

**Average filter:** smooths image by averaging neighbors.

**Median filter:** replaces pixel by neighborhood median.

Better for salt-and-pepper noise: Median filter, because it removes impulse noise without blurring edges heavily.

```python
# Convolutional low-pass (Average blur)
avg_kernel = np.ones((3,3))/9
avg_filtered = apply_kernel(sp_noisy, avg_kernel)

# Non-convolutional low-pass (Median filter)
median_filtered = cv2.medianBlur(sp_noisy, 3)
```

# Step 6: High-pass filters on original image

**Sharpen filter:** enhances edges while preserving structure.

**Edge detection filter:** highlights rapid intensity changes.

# Step 2: Frame count and FPS

```python
frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
fps = cap.get(cv2.CAP_PROP_FPS)

print("Total Frames:", frame_count)
print("FPS:", fps)
```

# Step 3: Salt-and-Pepper noise

```python
def add_salt_pepper_frame(frame, prob=0.02):
    noisy = frame.copy()

    salt = np.random.random(frame.shape[:2]) < prob/2
    pepper = np.random.random(frame.shape[:2]) < prob/2

    noisy[salt] = 255
    noisy[pepper] = 0

    return noisy


cap = cv2.VideoCapture(video_path)

width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

fourcc = cv2.VideoWriter_fourcc(*'mp4v')

noisy_out = cv2.VideoWriter(
    'noisy_output.mp4',
    fourcc,
    fps,
    (width, height)
)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    noisy = add_salt_pepper_frame(frame)
    noisy_out.write(noisy)

cap.release()
noisy_out.release()

print("Noisy video saved as noisy_output.mp4")
```

# Step 4: Filter noisy video (Median filter)

```python
cap = cv2.VideoCapture('noisy_output.mp4')

filtered_out = cv2.VideoWriter(
    'filtered_output.mp4',
    fourcc,
    fps,
    (width, height)
)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    filtered = cv2.medianBlur(frame, 3)
    filtered_out.write(filtered)

cap.release()
filtered_out.release()

print("Filtered video saved as filtered_output.mp4")
```
